# 01 — Hashing & Recursion Basics

Foundational DSA practice: digit manipulation, divisor-finding, hashing/frequency patterns, and a full recursion progression from basics to branching recursion (Fibonacci).

## 1. Digit Manipulation

Core idea: any digit-by-digit problem reduces to a `while num > 0` loop using `% 10` (extract last digit) and `// 10` (drop last digit).

### Count the number of digits in a number

```python
n = 56789
nums = n
count = 0

# Strip one digit at a time until nothing is left; each strip = one digit counted
while nums > 0:
    count += 1
    nums = nums // 10  # drop the last digit

print(count)
```

### Reverse a number
Extract the last digit with `% 10`, then rebuild the result by shifting it left (`* 10`) and adding the new digit — this naturally reverses the digit order.

```python
n = 67543
nums = n
result = 0

while nums > 0:
    ld = nums % 10               # last digit
    result = (result * 10) + ld  # shift result left, append last digit
    nums = nums // 10            # drop last digit from source

print(result)
```

### Armstrong number check
An Armstrong number equals the sum of its own digits each raised to the power of the digit count (e.g. 153 = 1³ + 5³ + 3³).

```python
n = 153
num = n
total = 0
nod = len(str(n))  # number of digits, used as the exponent

while num > 0:
    ld = num % 10
    total = total + (ld ** nod)  # add digit^nod to running total
    num = num // 10

print(total)  # if total == n, it's an Armstrong number
```

## 2. Finding Divisors

Two approaches to the same problem — brute force vs. the √n optimization. Divisors pair up around √n: if `i` divides `n`, so does `n/i`, so you only need to check up to `√n`.

### Brute force — O(n)
Checks every number up to `n` — simple but slow for large `n`.

```python
num = 27

result = []
for i in range(1, num // 2):
    if num % i == 0:
        result.append(i)
result.append(num)  # n is always a divisor of itself
print(result)
```

### Optimized — O(√n)
Only check up to `√n`; for every divisor `i` found, also record its pair `n // i` (skip the duplicate when `i == n // i`, i.e. at a perfect square).

```python
from math import sqrt
num = 36

result = []
for i in range(1, int(sqrt(num) + 1)):
    if num % i == 0:
        result.append(i)
        if i != num // i:
            result.append(num // i)  # the paired divisor
result.sort()
print(result)
```

## 3. Hashing & Frequency Counting

**Core idea:** precompute frequencies once into a hash structure (dict or array), then answer any number of lookups in O(1) each — instead of re-scanning the data for every query. This is the single biggest optimization jump in this notebook.

### Frequency of elements — dict with manual key check

```python
num = [1, 2, 3, 3, 9, 18, 2, 3, 4, 6, 9, 12, 18, 36]

nums = num
freq_dict = {}
for i in range(0, len(nums)):
    if nums[i] in freq_dict:
        freq_dict[nums[i]] += 1
    else:
        freq_dict[nums[i]] = 1
print(freq_dict)
```

### Same result, cleaner — using `dict.get(key, default)`
Avoids the manual if/else membership check above.

```python
num = [1, 2, 3, 3, 9, 18, 2, 3, 4, 6, 9, 12, 18, 36]

nums = num
hash_map = {}
for i in range(0, len(nums)):
    hash_map[nums[i]] = hash_map.get(nums[i], 0) + 1
print(hash_map)
```

**Hashing, in one line:** pre-storing values in a data structure (list/dict/set) so they can be fetched in O(1) later, instead of searching for them each time.

### The brute-force way to answer queries — O(n·m)
For every query value in `m`, rescan the *entire* array `n`. Works, but wasteful — this is the baseline the hashing approach below fixes.

```python
n = [1, 2, 3, 3, 9, 18, 2, 3, 4, 6, 9, 12, 18, 36]
m = [1, 4, 5, 9, 12, 42, 3, 45, 67, 8]

for num in n:
    count = 0
    for x in m:
        if num == x:
            count += 1
    print(num, count)
```

### Precomputed hash array — O(n + m)
Build a frequency array once from `n` (one O(n) pass), then answer each query in `m` with an O(1) lookup instead of rescanning. Direct-address table: index = value, so bounds-checking (`num < 0 or num > 10`) matters here since the array has fixed size.

```python
n = [1, 2, 3, 3, 9, 1, 2, 3, 4, 6, 9, 1, 1, 6]
m = [1, 4, 5, 9, 12, 42, 3, 45, 67, 8]

hash_list = [0] * 11  # fixed-size lookup table, index = value
for i in n:
    hash_list[i] += 1

print(hash_list)

for num in m:
    if num < 0 or num > 10:      # out of the table's range
        print(num, 0)
    else:
        print(num, hash_list[num])
```

### Same query-lookup pattern, using a dict instead of a fixed array
More flexible than the array version above — no bounds issue since dict keys aren't limited to a fixed index range.

```python
n = [1, 2, 3, 3, 9, 1, 2, 3, 4, 6, 9, 12, 14, 16, 67, 12, 12, 45, 45, 1, 1, 6]
m = [1, 4, 5, 9, 12, 42, 3, 45, 67, 8]

nums = n
freq_dict = {}
for i in range(0, len(n)):
    freq_dict[nums[i]] = freq_dict.get(nums[i], 0) + 1
print(freq_dict)

# Answer each query in m using the precomputed frequency dict — O(1) per lookup
new_dict = {}
for num in m:
    if num in freq_dict:
        new_dict[num] = freq_dict[num]
    else:
        new_dict[num] = 0
print(new_dict)
```

### Character frequency hashing
Same hashing pattern applied to characters instead of numbers — build a frequency map of the string, then look up only the characters of interest.

```python
s = "abudefghijklmnopqtuvwxyzABCDEFIJKLMNOPQRSTUVWXYZerysblktlngiBBSITUKSNVKSNBEKJ"
w = ["a", "b", "u", "y", "s", "F", "G", "H", "I", "K", "L"]

hash_list = {}
for ch in s:
    hash_list[ch] = hash_list.get(ch, 0) + 1
print(hash_list)

new_dict = {}
for ch in w:
    if ch in hash_list:
        new_dict[ch] = hash_list[ch]
    else:
        new_dict[ch] = 0
print(new_dict)
```

## 4. Recursion — Building Intuition

Starting from a plain function call, then working through pitfalls (global state), execution order (pre-order vs post-order), and accumulator patterns.

### Baseline — a plain function call (no recursion yet)

```python
def greet():
    print("good M0rn")

greet()
```

### Recursion with a global variable — and why this is fragile
Works, but relying on global state to track recursion depth is brittle: the state persists across calls unless manually reset, and it's harder to reason about than passing values as parameters (see the next examples).

```python
# Global variable to track recursion depth
_global_n = 0

def func():
    global _global_n            # declare intent to modify the global variable
    if _global_n == 4:
        return
    _global_n += 1
    func()                       # recursive call — note 'n' is NOT passed as a parameter
    print("AAA")

# Reset _global_n before the first call — important if func() is called more than once
_global_n = 0
func()
```

### Pre-order recursion — print BEFORE the recursive call
Executes top-down: each number prints immediately, then the function recurses deeper.

```python
def num(i, n):
    if i > n:
        return
    print(i)        # print happens before recursing further
    num(i + 1, n)

num(5, 28)
```

### Post-order recursion — print AFTER the recursive call
Executes bottom-up: the recursion goes all the way down first, then prints happen as the call stack unwinds — so output comes out in reverse order compared to the pre-order version above.

```python
def num(i, n):
    if i > n:
        return
    num(i + 1, n)
    print(i)         # print happens after recursing — on the way back up the stack

num(5, 28)
```

### Countdown — n down to 1

```python
def num(n):
    if n == 0:
        return
    print(n)
    num(n - 1)

num(28)
```

### Sum of 1 to N — accumulator passed as a parameter
The running total (`sum`) is carried forward through each call instead of being rebuilt from return values — similar in spirit to tail recursion.

```python
def sum_func(sum, i, n):
    if i > n:
        print(sum)
        return
    sum_func(sum + i, i + 1, n)

sum_func(0, 1, 10)
```

### Sum of 1 to N — using the return value instead
Classic recursive accumulation: each call adds its own `n` to the result of the smaller subproblem.

```python
def func(n):
    if n == 1:
        return 1
    return n + func(n - 1)

func(9)
```

### Factorial of N
Same return-value accumulation pattern as the sum above, with multiplication instead of addition.

```python
def func(n):
    if n == 1:
        return 1
    return n * func(n - 1)

func(5)
```

## 5. Recursion on Arrays & Strings

Two-pointer logic (converging `left`/`right` indices) translates directly into recursion — each recursive call is one 'step' that a loop would otherwise take.

### Reverse an array — iterative two-pointer

```python
def func(nums):
    left = 0
    right = len(nums) - 1
    while left < right:
        nums[left], nums[right] = nums[right], nums[left]  # swap ends
        left += 1
        right -= 1
    return nums

nums = list(map(int, input().split(",")))
func(nums)
```

### Reverse an array — recursive two-pointer
Same swap logic as above, but each swap is one recursive call instead of one loop iteration.

```python
def func(nums, left, right):
    if left < right:
        nums[left], nums[right] = nums[right], nums[left]
        func(nums, left + 1, right - 1)   # recurse inward
    return nums

nums = list(map(int, input().split(",")))
func(nums, 0, len(nums) - 1)
```

### Palindrome check — recursive two-pointer
Base case: pointers meet or cross (`left >= right`) → True. Otherwise compare the outer characters and recurse inward; any mismatch short-circuits to False.

```python
def func(s, left, right):
    if left >= right:
        return True
    if s[left] != s[right]:
        return False
    return func(s, left + 1, right - 1)

s = input()
func(s, 0, len(s) - 1)
```

## 6. Recursion — Fibonacci (Branching Recursion)

First example of **branching** recursion — two recursive calls per invocation (`fib(n-1) + fib(n-2)`), instead of the single-call recursion used everywhere else in this notebook. This version is exponential time, since it recomputes the same subproblems repeatedly — the natural next step in the learning path is memoizing this into O(n).

```python
class solution:
    def func(self, num):
        if num == 0 or num == 1:
            return num
        return self.func(num - 1) + self.func(num - 2)  # branches into two recursive calls

    def fib(self, n) -> int:
        answer = self.func(n)
        return answer

s = solution()
s.fib(9)
```
